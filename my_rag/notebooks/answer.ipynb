```python
from rank_bm25 import BM25Okapi

documents = [
    "Cancellation is free up to 48 hours before arrival",
    "Breakfast is served from 7 AM to 10 AM",
    "Deluxe rooms have a Nile view",
    "Guests can modify their booking before arrival"
]

tokenized_docs = [
    doc.lower().split()
    for doc in documents
]

bm25 = BM25Okapi(tokenized_docs)

query = "free cancellation"

scores = bm25.get_scores(query.lower().split())

for i in scores.argsort()[::-1]:
    print(f"Score: {scores[i]:.3f} | {documents[i]}")
```
