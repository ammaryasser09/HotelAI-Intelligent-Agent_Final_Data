```python
from pathlib import Path                                        
import re                                                     
from rank_bm25 import BM25Okapi                                  

ROOT = Path.cwd()                                                
while not (ROOT / "knowledge_base").exists() and ROOT != ROOT.parent:  
    ROOT = ROOT.parent                                           
KB_DIR = ROOT / "knowledge_base"                                
print(KB_DIR, KB_DIR.exists())                                   

def tokenize(text):                                              
    return re.findall(r"[a-z0-9]+", text.lower())              

names = []                                                     
corpus = []                                                     
for path in sorted(KB_DIR.glob("*.txt")):                        
    text = path.read_text(encoding="utf-8")                      
    names.append(path.name)                                      
    corpus.append(tokenize(text))                                
print(len(names), "files")                                       

bm25 = BM25Okapi(corpus, k1=1.5, b=0.75)                         

question = "What time is breakfast?"                          
scores = bm25.get_scores(tokenize(question))                  

results = []                                                    
for i in range(len(names)):                                     
    results.append((scores[i], names[i]))                       
results.sort(reverse=True)                                       

for score, name in results[:3]:                                
    print(round(score, 2), name)                                
```

```python
def bm25_search(question, top_k=3):                              
    scores = bm25.get_scores(tokenize(question))                 
    results = []                                                
    for i in range(len(names)):                                 
        results.append((round(scores[i], 2), names[i]))        
    results.sort(reverse=True)                                   
    return results[:top_k]                                       

questions = [                                                   
    "What time is breakfast?",
    "How much is the spa?",
    "Can I bring my dog?",
    "الفطار لحد إمتى؟",
]

for q in questions:                                            
    print(q)                                                
    for score, name in bm25_search(q):                          
        print("   ", score, name)                                
```
